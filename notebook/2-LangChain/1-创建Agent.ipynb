# 1.加载环境变量

```python
from dashscope.common.env import api_key
from dotenv import load_dotenv
import os
load_dotenv()
api_key=os.getenv("ALIYUN")
```

# 2.定义工具

```python
from langchain.tools import tool

@tool
def get_weather(location:str) -> str:
    """
    获取天气
    :param location:
    :return:
    """
    return f"今天{location}是晴天"
```

# 3.定义agent

```python
from langchain.agents import create_agent
from langchain_community.chat_models import ChatTongyi

agent = create_agent(
    model=ChatTongyi(
        model="qwen-max",
        api_key=api_key
    ),
    tools=[get_weather],
)
```

# 4.发起调用

```python
response = agent.invoke({"messages": [{"role": "user", "content": "今天深圳天气如何"}, ]})
# print(response)
# print(type(response))
# print(response['messages'])
for message in response['messages']:
    print('='*20,message,'='*20)
```
