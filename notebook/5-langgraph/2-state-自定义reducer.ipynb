```python
from typing import TypedDict,Annotated
from IPython.display import Image, display
from langgraph.constants import START, END
from langgraph.graph import StateGraph
from operator import add

#=================定义Reducer函数================
def add_reducer(old,new):
    return old + new
#关于列表相加：
# a=[1,2,3]
# b=[4,5,6]
# print(a+b) ——————>[1,2,3,4,5,6]

#======================1.State======================
class CustomReducerState(TypedDict):
    nodes:Annotated[list[str],add_reducer]
    count:int

#======================2.Node======================
def node_1(state: CustomReducerState):
    print(f"node_1接收到了state：{state}")
    return {
        "count":state['count']+1,
        "nodes":['node_1']
    }

def node_2(state:CustomReducerState):
    print(f"node_2接收到了state：{state}")
    return {
        "count":state['count']+1,
        "nodes":['node_2']
    }

#======================3.Edge连接Node形成Graph======================
graph_builder=StateGraph(CustomReducerState)
graph_builder.add_node("node1",node_1)
graph_builder.add_node("node2",node_2)
graph_builder.add_edge(START,"node1")
graph_builder.add_edge("node1","node2")
graph_builder.add_edge("node2",END)

graph = graph_builder.compile()

# ======================4.执行Graph======================
result = graph.invoke({"count": 0, "nodes": []})
print(result)
```
