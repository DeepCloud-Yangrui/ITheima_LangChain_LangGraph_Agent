# 1、ollama Embedding

```python
from langchain_ollama import OllamaEmbeddings

# 1.创建embedding模型
ollama_embeddings=OllamaEmbeddings(
    model="qwen3-embedding:4b",
    dimensions=1024,
)

# 2.向量化单条文本
text="我爱上班"
vector = ollama_embeddings.embed_query(text)
print(type(vector))
print(vector)
```

```python
# 批量向量化
texts=["我要躺平","我爱工作","拒绝加班"]
vectors = ollama_embeddings.embed_documents(texts)
print(f"\n批量向量化{len(vectors)}条，每条维度是：{len(vectors[0])}")
print(vectors)
```

```python
import math
import numpy as np

# 3.计算文本相似度

# 计算余弦相似度的函数
def cosine_similarity(vec1, vec2):
    dot = np.dot(vec1, vec2)
    return dot / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

for i,v in enumerate(vectors):
    print(f"文本{text}与文本{texts[i]}的相似度是：{cosine_similarity(vector,v)}")
```

# 2、DashScope Embedding

```python
from dotenv import load_dotenv
from langchain_community.embeddings import DashScopeEmbeddings

load_dotenv()
# 1.创建embedding模型，只需要指定模型名就可以，dashScopeEmbedding会自动取.env文件中的API_KEY
dash_scope_embeddings = DashScopeEmbeddings(model="text-embedding-v4", )

# 向量化单条文本
text="我爱上班"
vector = dash_scope_embeddings.embed_query(text)
print(vector)
```

```python
# 批量向量化
texts=["我要躺平","我爱工作","拒绝加班"]
vectors = dash_scope_embeddings.embed_documents(texts)
def cosine_similarity(vec1, vec2):
    dot = np.dot(vec1, vec2)
    return dot / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

for i,v in enumerate(vectors):
    print(f"文本{text}与文本{texts[i]}的相似度是：{cosine_similarity(vector,v)}")
```
